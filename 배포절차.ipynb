 **Streamlit Community Cloud 기준**으로 가장 간단한 배포 절차



1. 프로젝트 폴더 준비
```text
my_streamlit_app/
├─ app.py
└─ requirements.txt
```

`app.py`
```python
import streamlit as st

st.title("나의 첫 Streamlit 배포")
st.write("Hello Streamlit!")
```

`requirements.txt`
```text
streamlit
pandas
plotly
```

2. 로컬에서 먼저 실행 확인
```bash
streamlit run app.py
```

브라우저에서 정상 실행되는지 확인하면 됨.

3. GitHub에 프로젝트 업로드

GitHub에서 새 Repository 생성 후 프로젝트 파일을 올림.

예:
```text
GitHub
└─ my_streamlit_app
   ├─ app.py
   └─ requirements.txt
```

터미널에서 올린다면:

```bash
git init
git add .
git commit -m "first streamlit app"
git branch -M main
git remote add origin https://github.com/사용자명/my_streamlit_app.git
git push -u origin main
```

4. Streamlit Community Cloud 접속

[Streamlit Community Cloud](https://share.streamlit.io/?utm_source=chatgpt.com)

GitHub 계정으로 로그인함.

5. `Create app` 또는 `New app` 선택

보통 다음 3가지를 지정함.

```text
Repository
사용자명/my_streamlit_app

Branch
main

Main file path
app.py
```

그리고 **Deploy** 버튼을 누르면 됨.

6. 배포가 완료되면 URL이 생성됨.

예:
```text
https://my-streamlit-app.streamlit.app
```

### 가장 중요한 `requirements.txt`

내 PC에서는 되는데 배포 서버에서 안 되는 경우 대부분 여기 문제임.

예를 들어:

```python
import streamlit as st
import pandas as pd
import plotly.express as px
from openai import OpenAI
```

를 사용했다면:

```text
streamlit
pandas
plotly
openai
```

가 필요함.

---

### OpenAI API Key를 사용하는 경우

API Key를 `app.py`에 직접 작성하면 안 됨.

```python
# 이렇게 하면 안 됨
client = OpenAI(api_key="sk-xxxx")
```

Streamlit에서는 Secrets를 사용함.

Streamlit Cloud의

```text
App
→ Settings
→ Secrets
```

에 다음처럼 등록:

```toml
OPENAI_API_KEY = "sk-xxxx"
```

코드에서는:

```python
import streamlit as st
from openai import OpenAI

client = OpenAI(
    api_key=st.secrets["OPENAI_API_KEY"]
)
```

---


```text
① Streamlit 코드 작성
        ↓
② requirements.txt 작성
        ↓
③ GitHub에 업로드
        ↓
④ Streamlit Community Cloud 로그인
        ↓
⑤ GitHub Repository 선택
        ↓
⑥ app.py 선택
        ↓
⑦ Deploy
        ↓
⑧ xxx.streamlit.app 주소 생성
```

즉, 핵심은 **`내 PC → GitHub → Streamlit Cloud → 웹사이트`** 구조임.
